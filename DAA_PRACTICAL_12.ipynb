```python

from itertools import permutations

def tsp(graph, start):
    n = len(graph)
    cities = [i for i in range(n) if i != start]

    min_cost = float('inf')
    best_path = None

    for path in permutations(cities):
        cost = 0
        current = start

        for city in path:
            cost += graph[current][city]
            current = city

        cost += graph[current][start]

        if cost < min_cost:
            min_cost = cost
            best_path = (start,) + path + (start,)

    print("Shortest path:", " -> ".join(map(str, best_path)))
    print("Minimum cost:", min_cost)


graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

tsp(graph, 0)
```
