```python

# Floyd-Warshall Algorithm

INF = float('inf')

# Adjacency matrix of the graph
graph = [
    [0,   5,   INF, 10],
    [INF, 0,   3,   INF],
    [INF, INF, 0,   1],
    [INF, INF, INF, 0]
]

n = len(graph)

# Floyd-Warshall algorithm
dist = [row[:] for row in graph]

for k in range(n):
    for i in range(n):
        for j in range(n):
            dist[i][j] = min(
                dist[i][j],
                dist[i][k] + dist[k][j]
            )

# Print the shortest distance matrix
print("Shortest distance matrix:")

for i in range(n):
    for j in range(n):
        if dist[i][j] == INF:
            print("INF", end="   ")
        else:
            print(dist[i][j], end="     ")
    print()
```
