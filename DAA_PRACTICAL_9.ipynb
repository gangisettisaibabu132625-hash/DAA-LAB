```python
graph = [
    ['A', 'B', 3],
    ['B', 'C', 6],
    ['C', 'D', 2],
    ['D', 'A', 4]
]

v = ['A', 'B', 'C', 'D', 'E']

key = {x: float('inf') for x in v}
parent = {x: None for x in v}
visited = {x: False for x in v}

key['C'] = 0

for _ in v:

    u = min((x for x in v if not visited[x]), key=lambda x: key[x])
    visited[u] = True

    for a, b, w in graph:
        if a == u:
            x = b
        elif b == u:
            x = a
        else:
            continue

        if not visited[x] and w < key[x]:
            key[x] = w
            parent[x] = u

total = 0

print("MST Edges:")

for x in v:
    if parent[x]:
        print(parent[x], "-", x, ":", key[x])
        total += key[x]

print("Minimum Cost:", total)
```
